```python
print(" AKSHAYA PET SHOP")
print("phone:1234567890")
print("________________")
total_amount = 0

while True:

    Pets = input("\nWhich pet do you want? (Dog, Birds, Aquarium_fish, Small_pet): ")

    if Pets == "Dog":
        print("pomeranian, labrador retriever, Beagle, golden retriever")
        choice = input("Choose the breed: ")

        if choice == "pomeranian":
            price = 10000
        elif choice == "labrador retriever":
            price = 25000
        elif choice == "Beagle":
            price = 30000
        elif choice == "golden retriever":
            price = 100000
        else:
            print("Invalid breed")
            continue

    elif Pets == "Birds":
        print("finches, budgies, lovebirds, cockatiels")
        choice = input("Choose the breed: ")

        if choice == "finches":
            price = 1000
        elif choice == "budgies":
            price = 2000
        elif choice == "lovebirds":
            price = 3000
        elif choice == "cockatiels":
            price = 4000
        else:
            print("Invalid breed")
            continue

    elif Pets == "Aquarium_fish":
        print("gold fish, guppies, betta, angel fish, cichilds")
        choice = input("Choose a breed: ")

        if choice == "gold fish":
            price = 50
        elif choice == "guppies":
            price = 80
        elif choice == "betta":
            price = 100
        elif choice == "angel fish":
            price = 250
        elif choice == "cichilds":
            price = 500
        else:
            print("Invalid breed")
            continue

    elif Pets == "Small_pet":
        print("hamsters, mice, rabbits, guinea pigs")
        choice = input("Choose a breed: ")

        if choice == "hamsters":
            price = 500
        elif choice == "mice":
            price = 1000
        elif choice == "rabbits":
            price = 2000
        elif choice == "guinea pigs":
            price = 2000
        else:
            print("Invalid breed")
            continue

    else:
        print("Invalid pet category")
        continue

    
    print("Price of one", choice, ":", price)

    
    quantity = int(input("How many do you want? "))

   
    amount = price * quantity

    print("Amount for", quantity, choice, ":", amount)

   
    total_amount += amount

    print("Current total:", total_amount)

    
    more = input("Do you want to buy another pet? (yes/no): ")

    if more == "no":
        break

print("\n-------------------------")
print("Final total amount:", total_amount)
print("-------------------------")
print("Thank you for shopping!")
```
